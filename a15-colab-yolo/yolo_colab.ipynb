```python
# Celda 3: carga YOLO26 (los pesos se descargan solos) y analiza la foto
from ultralytics import YOLO
modelo = YOLO("yolo26n.pt")
r = modelo("foto.jpg", save=True)
```

```python
# Celda 4: lista cada objeto con su confianza (de 0 a 1)
for caja in r[0].boxes:
    print(modelo.names[int(caja.cls)], round(float(caja.conf), 2))
```

```python
# Celda 5: muestra la foto con los recuadros (plot() la entrega en BGR; se pasa a RGB)
from PIL import Image
Image.fromarray(r[0].plot()[:, :, ::-1].copy())
```
