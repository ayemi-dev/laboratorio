# ¿Qué te tocó hoy en Colab?

Google no publica las especificaciones de la versión gratuita de Colab: su FAQ dice que los límites cambian con el tiempo. Esta celda te dice qué máquina te tocó **a ti, hoy**.

Ejecuta la celda de abajo (botón ▶ o `Ctrl+Enter`). Cada línea empieza con `!` para que el cuaderno la mande al sistema.

```python
!nproc
!free -h
!df -h /
!nvidia-smi
```

## Cómo leer el resultado

| Línea | Qué te dice |
|---|---|
| `!nproc` | Cuántos procesadores tiene tu máquina |
| `!free -h` | La memoria: fila `Mem`, columnas `total` y `available` |
| `!df -h /` | El disco: columnas `Size` (tamaño) y `Avail` (libre) |
| `!nvidia-smi` | Qué GPU te tocó y su memoria |

Si `!nvidia-smi` da error, es normal: tu entorno de ejecución no tiene GPU. Para pedirla, cambia el tipo de entorno de ejecución y vuelve a ejecutar la celda. Cuando ya no la uses, regrésalo a ninguno.
